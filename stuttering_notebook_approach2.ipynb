````python
import numpy as np
import librosa
import json
import os
import torch
from google import genai

def summarize_vector(vec, name="", max_len=20):
    """
    Returns a small summary of a long numeric vector.
    Computes mean, std, min, max and (optionally) first few elements.
    """
    if torch.is_tensor(vec):
        vec = vec.detach().cpu().numpy()
    vec = np.array(vec).astype(float).flatten()
    summary = {
        "mean": float(np.mean(vec)),
        "std": float(np.std(vec)),
        "min": float(np.min(vec)),
        "max": float(np.max(vec)),
    }
    # keep the first few numbers to give the LLM a sense of shape
    if len(vec) > 0:
        summary["head"] = np.round(vec[:max_len], 3).tolist()
    if name:
        return {name: summary}
    return summary


def build_feature_summary(audio_path, mfcc_npy=None, wav2vec_npy=None, prosody_npy=None):
    """
    Combines multiple sources (audio + .npy feature files) into
    one compact JSON-serializable dict for Gemini input.
    """
    summary = {}

    # ---- Audio-level descriptors (librosa) ----
    try:
        y, sr = librosa.load(audio_path, sr=16000)
        tempo, _ = librosa.beat.beat_track(y=y, sr=sr)
        pitch = librosa.yin(y, fmin=50, fmax=400)
        energy = np.mean(y ** 2)
        summary["audio_summary"] = {
            "sample_rate": sr,
            "duration_sec": round(len(y) / sr, 2),
            "mean_pitch": round(float(np.mean(pitch)), 2),
            "pitch_std": round(float(np.std(pitch)), 2),
            "mean_energy": round(float(energy), 5),
            "tempo": round(float(tempo), 2),
        }
    except Exception as e:
        summary["audio_summary"] = {"error": f"Audio analysis failed: {e}"}

    # ---- MFCC ----
    if mfcc_npy and os.path.exists(mfcc_npy):
        try:
            mfcc = np.load(mfcc_npy)
            summary.update(summarize_vector(mfcc.mean(axis=1), name="mfcc_mean"))
            summary.update(summarize_vector(mfcc.std(axis=1), name="mfcc_std"))
        except Exception as e:
            summary["mfcc_error"] = str(e)

    # ---- wav2vec2 ----
    if wav2vec_npy and os.path.exists(wav2vec_npy):
        try:
            w2v = np.load(wav2vec_npy)
            summary.update(summarize_vector(w2v.mean(axis=0), name="wav2vec2_mean"))
            summary.update(summarize_vector(w2v.std(axis=0), name="wav2vec2_std"))
        except Exception as e:
            summary["wav2vec2_error"] = str(e)

    # ---- prosody ----
    if prosody_npy and os.path.exists(prosody_npy):
        try:
            pros = np.load(prosody_npy, allow_pickle=True)
            pros_summary = {}
    
            if isinstance(pros, dict):
                # already a dict of features
                for k, v in pros.items():
                    pros_summary[k] = summarize_vector(v, name=k)
            elif pros.dtype == object and isinstance(pros.item(), dict):
                # object array containing dict
                pros = pros.item()
                for k, v in pros.items():
                    pros_summary[k] = summarize_vector(v, name=k)
            else:
                # assume plain numeric array -> summarize directly
                pros_summary["prosody_array"] = summarize_vector(pros, name="prosody_array")
    
            summary["prosody_summary"] = pros_summary
    
        except Exception as e:
            summary["prosody_error"] = str(e)

        return summary




def build_json_for_llm(audio_path, mfcc_npy=None, wav2vec_npy=None, prosody_npy=None, save_path=None):
    """
    Returns JSON string ready to send to Gemini (or saves to disk).
    """
    summary = build_feature_summary(audio_path, mfcc_npy, wav2vec_npy, prosody_npy)
    json_str = json.dumps(summary, indent=2)
    if save_path:
        with open(save_path, "w") as f:
            f.write(json_str)
    return json_str
````

````python
import os, json, time, random, re
import pandas as pd
from glob import glob
from google import genai
from tqdm import tqdm
import numpy as np
import librosa
import torch

# ===========================
# CONFIGURATION
# ===========================
API_KEY = ""                # Insert API_KEY here
DATA_DIR = r"C:\Tampu\stuttering\data_sep28k\clips_3s"
FEATURE_DIR = r"C:\Tampu\stuttering\data_sep28k\features"
OUTPUT_CSV = "gemini_prediction_reasoning_prompt_updated.csv"
SELECTED_FILES_CSV = "selected_files.csv"
MAX_SAMPLES = 100
SEED = 21  # reproducible random seed

client = genai.Client(api_key=API_KEY)

# ======================================================
# FEATURE BUILDERS
# ======================================================
def summarize_vector(vec, name="", max_len=20):
    if torch.is_tensor(vec):
        vec = vec.detach().cpu().numpy()
    vec = np.array(vec).astype(float).flatten()
    summary = {
        "mean": float(np.mean(vec)),
        "std": float(np.std(vec)),
        "min": float(np.min(vec)),
        "max": float(np.max(vec)),
    }
    if len(vec) > 0:
        summary["head"] = np.round(vec[:max_len], 3).tolist()
    if name:
        return {name: summary}
    return summary


def build_feature_summary(audio_path, mfcc_npy=None, wav2vec_npy=None, prosody_npy=None):
    summary = {}
    try:
        y, sr = librosa.load(audio_path, sr=16000)
        tempo, _ = librosa.beat.beat_track(y=y, sr=sr)
        pitch = librosa.yin(y, fmin=50, fmax=400)
        energy = np.mean(y ** 2)
        summary["audio_summary"] = {
            "sample_rate": sr,
            "duration_sec": round(len(y) / sr, 2),
            "mean_pitch": round(float(np.mean(pitch)), 2),
            "pitch_std": round(float(np.std(pitch)), 2),
            "mean_energy": round(float(energy), 5),
            "tempo": round(float(tempo), 2),
        }
    except Exception as e:
        summary["audio_summary"] = {"error": f"Audio analysis failed: {e}"}

    if mfcc_npy and os.path.exists(mfcc_npy):
        try:
            mfcc = np.load(mfcc_npy)
            summary.update(summarize_vector(mfcc.mean(axis=1), name="mfcc_mean"))
            summary.update(summarize_vector(mfcc.std(axis=1), name="mfcc_std"))
        except Exception as e:
            summary["mfcc_error"] = str(e)

    if wav2vec_npy and os.path.exists(wav2vec_npy):
        try:
            w2v = np.load(wav2vec_npy)
            summary.update(summarize_vector(w2v.mean(axis=0), name="wav2vec2_mean"))
            summary.update(summarize_vector(w2v.std(axis=0), name="wav2vec2_std"))
        except Exception as e:
            summary["wav2vec2_error"] = str(e)

    # ---- prosody ----
    if prosody_npy and os.path.exists(prosody_npy):
        try:
            pros = np.load(prosody_npy, allow_pickle=True)
            pros_summary = {}
            if isinstance(pros, dict):
                for k, v in pros.items():
                    pros_summary[k] = summarize_vector(v, name=k)
            elif pros.dtype == object and isinstance(pros.item(), dict):
                pros = pros.item()
                for k, v in pros.items():
                    pros_summary[k] = summarize_vector(v, name=k)
            else:
                pros_summary["prosody_array"] = summarize_vector(pros, name="prosody_array")
            summary["prosody_summary"] = pros_summary
        except Exception as e:
            summary["prosody_error"] = str(e)

    return summary


def build_json_for_llm(audio_path, mfcc_npy=None, wav2vec_npy=None, prosody_npy=None, save_path=None):
    summary = build_feature_summary(audio_path, mfcc_npy, wav2vec_npy, prosody_npy)
    json_str = json.dumps(summary, indent=2)
    if save_path:
        with open(save_path, "w") as f:
            f.write(json_str)
    return json_str


# ======================================================
# PROMPT (define prompt_reasoning_updates)
# ======================================================
prompt = prompt_reasoning_updated + examples

# ======================================================
# RANDOM SAMPLE SELECTION
# ======================================================
random.seed(SEED)
all_wavs = glob(os.path.join(DATA_DIR, "**", "*.wav"), recursive=True)
print(f"Found {len(all_wavs)} total WAV files under {DATA_DIR}")

wav_files = random.sample(all_wavs, min(MAX_SAMPLES, len(all_wavs)))
pd.DataFrame({"file": wav_files}).to_csv(SELECTED_FILES_CSV, index=False)
print(f"Randomly (reproducibly) selected {len(wav_files)} files using seed={SEED}.")
print(f"Saved list to {SELECTED_FILES_CSV}")

# ======================================================
# LOOP THROUGH SAMPLES
# ======================================================
results = []

for wav_path in tqdm(wav_files, desc="Processing samples"):
    try:
        wav_file = os.path.basename(wav_path)
        base = os.path.splitext(wav_file)[0]

        mfcc_npy = os.path.join(FEATURE_DIR, f"{base}_mfcc.npy")
        w2v_npy = os.path.join(FEATURE_DIR, f"{base}_wav2vec.npy")
        pros_npy = os.path.join(FEATURE_DIR, f"{base}_prosody.npy")

        feature_json = build_json_for_llm(wav_path, mfcc_npy, w2v_npy, pros_npy)
        uploaded_audio = client.files.upload(file=wav_path)

        # -----------------------------
        # Query Gemini and parse output
        # -----------------------------
        response = client.models.generate_content(
            model="gemini-2.5-flash",
            contents=[
                {"role": "user", "parts": [
                    {"text": prompt},
                    {"file_data": {"file_uri": uploaded_audio.uri}},
                    {"text": feature_json}
                ]}
            ]
        )

        # ---- Clean & parse JSON output ----
        raw_text = response.text.strip()

        # --- Step 1: Split reasoning and JSON parts ---
        reasoning_text, json_text = "", ""
        split_index = raw_text.find("[")
        if split_index != -1:
            reasoning_text = raw_text[:split_index].strip()
            json_text = raw_text[split_index:].strip()
        else:
            reasoning_text = raw_text
            json_text = "[]"

        # --- Step 2: Remove markdown fences if any ---
        if json_text.startswith("```"):
            json_text = re.sub(r"^```[a-zA-Z]*", "", json_text)
            json_text = re.sub(r"```$", "", json_text).strip()

       # --- Step 3: Parse JSON safely ---
        try:
            # --- Strong fence cleanup ---
            json_text = re.sub(r"^.*?```json", "", json_text, flags=re.DOTALL)
            json_text = re.sub(r"```$", "", json_text, flags=re.MULTILINE)
            json_text = json_text.replace("```json", "").replace("```", "").strip()

            try:
                output_json = json.loads(json_text)
            except json.JSONDecodeError as e:
                if "Extra data" in str(e):
                    # truncate after first closing bracket if stray text exists
                    closing_index = json_text.find("]") + 1
                    json_text = json_text[:closing_index]
                    output_json = json.loads(json_text)
                else:
                    raise
            if isinstance(output_json, dict):
                output_json = [output_json]

        except json.JSONDecodeError as e:
            print(f"[Warning] Could not parse JSON for {wav_file}: {e}")
            print("Raw response:\n", response.text)
            output_json = []

        # --- Step 4: Save reasoning trace for inspection ---
        reasoning_path = "gemini_reasoning_log.txt"
        with open(reasoning_path, "a", encoding="utf-8") as f:
            f.write(f"\n===== {wav_file} =====\n")
            f.write(reasoning_text + "\n")

        # --- Step 5: Append structured results ---
        for entry in output_json:
            results.append({
                "file": wav_file,
                "path": wav_path,
                "event_type": entry.get("event_type"),
                "start_time": entry.get("start_time"),
                "end_time": entry.get("end_time"),
                "confidence": entry.get("confidence"),
                "reasoning": reasoning_text  # include reasoning in CSV
            })

        time.sleep(0.2)  # gentle rate limit

    except Exception as e:
        print(f"[Error] {wav_path}: {e}")
        continue

# ======================================================
# SAVE RESULTS
# ======================================================
df = pd.DataFrame(results)
df.to_csv(OUTPUT_CSV, index=False)
print(f"\n Done! Processed {len(wav_files)} samples, saved predictions to {OUTPUT_CSV}")

````

````python
import os, json, time, random, re
import pandas as pd
from glob import glob

from google import genai
from tqdm import tqdm
import numpy as np
import librosa
import torch

# ===========================
# CONFIGURATION
# ===========================
API_KEY = ""         # INSERT API_KEY here
DATA_DIR = r"C:\Tampu\stuttering\data_sep28k\clips_3s"
FEATURE_DIR = r"C:\Tampu\stuttering\data_sep28k\features"
OUTPUT_CSV = "gemini_prediction_reasoning_prompt_updated_try_2.csv"
SELECTED_FILES_CSV = "selected_files.csv"
MAX_SAMPLES = 100
SEED = 21  # reproducible random seed

client = genai.Client(api_key=API_KEY)

# ======================================================
# FEATURE BUILDERS
# ======================================================
def summarize_vector(vec, name="", max_len=20):
    if torch.is_tensor(vec):
        vec = vec.detach().cpu().numpy()
    vec = np.array(vec).astype(float).flatten()
    summary = {
        "mean": float(np.mean(vec)),
        "std": float(np.std(vec)),
        "min": float(np.min(vec)),
        "max": float(np.max(vec)),
    }
    if len(vec) > 0:
        summary["head"] = np.round(vec[:max_len], 3).tolist()
    if name:
        return {name: summary}
    return summary


def build_feature_summary(audio_path, mfcc_npy=None, wav2vec_npy=None, prosody_npy=None):
    summary = {}
    try:
        y, sr = librosa.load(audio_path, sr=16000)
        tempo, _ = librosa.beat.beat_track(y=y, sr=sr)
        pitch = librosa.yin(y, fmin=50, fmax=400)
        energy = np.mean(y ** 2)
        summary["audio_summary"] = {
            "sample_rate": sr,
            "duration_sec": round(len(y) / sr, 2),
            "mean_pitch": round(float(np.mean(pitch)), 2),
            "pitch_std": round(float(np.std(pitch)), 2),
            "mean_energy": round(float(energy), 5),
            "tempo": round(float(tempo), 2),
        }
    except Exception as e:
        summary["audio_summary"] = {"error": f"Audio analysis failed: {e}"}

    if mfcc_npy and os.path.exists(mfcc_npy):
        try:
            mfcc = np.load(mfcc_npy)
            summary.update(summarize_vector(mfcc.mean(axis=1), name="mfcc_mean"))
            summary.update(summarize_vector(mfcc.std(axis=1), name="mfcc_std"))
        except Exception as e:
            summary["mfcc_error"] = str(e)

    if wav2vec_npy and os.path.exists(wav2vec_npy):
        try:
            w2v = np.load(wav2vec_npy)
            summary.update(summarize_vector(w2v.mean(axis=0), name="wav2vec2_mean"))
            summary.update(summarize_vector(w2v.std(axis=0), name="wav2vec2_std"))
        except Exception as e:
            summary["wav2vec2_error"] = str(e)

    # ---- prosody ----
    if prosody_npy and os.path.exists(prosody_npy):
        try:
            pros = np.load(prosody_npy, allow_pickle=True)
            pros_summary = {}
            if isinstance(pros, dict):
                for k, v in pros.items():
                    pros_summary[k] = summarize_vector(v, name=k)
            elif pros.dtype == object and isinstance(pros.item(), dict):
                pros = pros.item()
                for k, v in pros.items():
                    pros_summary[k] = summarize_vector(v, name=k)
            else:
                pros_summary["prosody_array"] = summarize_vector(pros, name="prosody_array")
            summary["prosody_summary"] = pros_summary
        except Exception as e:
            summary["prosody_error"] = str(e)

    return summary


def build_json_for_llm(audio_path, mfcc_npy=None, wav2vec_npy=None, prosody_npy=None, save_path=None):
    summary = build_feature_summary(audio_path, mfcc_npy, wav2vec_npy, prosody_npy)
    json_str = json.dumps(summary, indent=2)
    if save_path:
        with open(save_path, "w") as f:
            f.write(json_str)
    return json_str


# ======================================================
# PROMPT (define prompt_reasoning_updates)
# ======================================================
prompt = prompt_reasoning_updated + examples

# ======================================================
# RANDOM SAMPLE SELECTION
# ======================================================
random.seed(SEED)
all_wavs = glob(os.path.join(DATA_DIR, "**", "*.wav"), recursive=True)
print(f"Found {len(all_wavs)} total WAV files under {DATA_DIR}")

wav_files = random.sample(all_wavs, min(MAX_SAMPLES, len(all_wavs)))
pd.DataFrame({"file": wav_files}).to_csv(SELECTED_FILES_CSV, index=False)
print(f"Randomly (reproducibly) selected {len(wav_files)} files using seed={SEED}.")
print(f"Saved list to {SELECTED_FILES_CSV}")

# ======================================================
# LOOP THROUGH SAMPLES
# ======================================================
results = []

for wav_path in tqdm(wav_files, desc="Processing samples"):
    try:
        wav_file = os.path.basename(wav_path)
        base = os.path.splitext(wav_file)[0]

        mfcc_npy = os.path.join(FEATURE_DIR, f"{base}_mfcc.npy")
        w2v_npy = os.path.join(FEATURE_DIR, f"{base}_wav2vec.npy")
        pros_npy = os.path.join(FEATURE_DIR, f"{base}_prosody.npy")

        feature_json = build_json_for_llm(wav_path, mfcc_npy, w2v_npy, pros_npy)
        uploaded_audio = client.files.upload(file=wav_path)

        # -----------------------------
        # Query Gemini and parse output
        # -----------------------------
        response = client.models.generate_content(
            model="gemini-2.5-flash",
            contents=[
                {"role": "user", "parts": [
                    {"text": prompt},
                    {"file_data": {"file_uri": uploaded_audio.uri}},
                    {"text": feature_json}
                ]}
            ]
        )

        # ---- Clean & parse JSON output ----
        raw_text = response.text.strip()

        # --- Step 1: Split reasoning and JSON parts ---
        reasoning_text, json_text = "", ""
        split_index = raw_text.find("[")
        if split_index != -1:
            reasoning_text = raw_text[:split_index].strip()
            json_text = raw_text[split_index:].strip()
        else:
            reasoning_text = raw_text
            json_text = "[]"

        # --- Step 2: Remove markdown fences if any ---
        if json_text.startswith("```"):
            json_text = re.sub(r"^```[a-zA-Z]*", "", json_text)
            json_text = re.sub(r"```$", "", json_text).strip()

       # --- Step 3: Parse JSON safely ---
        try:
            # --- Strong fence cleanup ---
            json_text = re.sub(r"^.*?```json", "", json_text, flags=re.DOTALL)
            json_text = re.sub(r"```$", "", json_text, flags=re.MULTILINE)
            json_text = json_text.replace("```json", "").replace("```", "").strip()

            try:
                output_json = json.loads(json_text)
            except json.JSONDecodeError as e:
                if "Extra data" in str(e):
                    # truncate after first closing bracket if stray text exists
                    closing_index = json_text.find("]") + 1
                    json_text = json_text[:closing_index]
                    output_json = json.loads(json_text)
                else:
                    raise
            if isinstance(output_json, dict):
                output_json = [output_json]

        except json.JSONDecodeError as e:
            print(f"[Warning] Could not parse JSON for {wav_file}: {e}")
            print("Raw response:\n", response.text)
            output_json = []

        # --- Step 4: Save reasoning trace for inspection ---
        reasoning_path = "gemini_reasoning_log.txt"
        with open(reasoning_path, "a", encoding="utf-8") as f:
            f.write(f"\n===== {wav_file} =====\n")
            f.write(reasoning_text + "\n")

        # --- Step 5: Append structured results ---
        for entry in output_json:
            results.append({
                "file": wav_file,
                "path": wav_path,
                "event_type": entry.get("event_type"),
                "start_time": entry.get("start_time"),
                "end_time": entry.get("end_time"),
                "confidence": entry.get("confidence"),
                "reasoning": reasoning_text  # include reasoning in CSV
            })

        time.sleep(0.2)  # gentle rate limit

    except Exception as e:
        print(f"[Error] {wav_path}: {e}")
        continue

# ======================================================
# SAVE RESULTS
# ======================================================
df = pd.DataFrame(results)
df.to_csv(OUTPUT_CSV, index=False)
print(f"\n Done! Processed {len(wav_files)} samples, saved predictions to {OUTPUT_CSV}")

````

````python
import os, json, re, time, random
import pandas as pd
from glob import glob
from tqdm import tqdm
import numpy as np
import librosa
import torch
from concurrent.futures import ThreadPoolExecutor, as_completed
from google import genai

# ===========================
# CONFIGURATION
# ===========================
API_KEY = ""                    # INSERT API_KEY here
DATA_DIR = r"C:\Tampu\stuttering\data_sep28k\clips_3s"
FEATURE_DIR = r"C:\Tampu\stuttering\data_sep28k\features"
OUTPUT_DIR = "results_fast"
MAX_SAMPLES = 100
SEED = 42
NUM_THREADS = 6   # safe parallelism
MODEL = "gemini-2.5-flash" 

os.makedirs(OUTPUT_DIR, exist_ok=True)
device = "cuda" if torch.cuda.is_available() else "cpu"
print(f"Using device: {device}, threads={NUM_THREADS}, model={MODEL}")

client = genai.Client(api_key=API_KEY)

# ======================================================
# FEATURE BUILDERS
# ======================================================
def summarize_vector(vec, name="", max_len=20):
    if torch.is_tensor(vec):
        vec = vec.detach().cpu().numpy()
    vec = np.array(vec).astype(float).flatten()
    summary = {
        "mean": float(np.mean(vec)),
        "std": float(np.std(vec)),
        "min": float(np.min(vec)),
        "max": float(np.max(vec)),
    }
    if len(vec) > 0:
        summary["head"] = np.round(vec[:max_len], 3).tolist()
    if name:
        return {name: summary}
    return summary


def build_feature_summary(audio_path, mfcc_npy=None, wav2vec_npy=None, prosody_npy=None):
    summary = {}
    try:
        y, sr = librosa.load(audio_path, sr=16000)
        tempo, _ = librosa.beat.beat_track(y=y, sr=sr)
        pitch = librosa.yin(y, fmin=50, fmax=400)
        energy = np.mean(y ** 2)
        summary["audio_summary"] = {
            "sample_rate": sr,
            "duration_sec": round(len(y) / sr, 2),
            "mean_pitch": round(float(np.mean(pitch)), 2),
            "pitch_std": round(float(np.std(pitch)), 2),
            "mean_energy": round(float(energy), 5),
            "tempo": round(float(tempo), 2),
        }
    except Exception as e:
        summary["audio_summary"] = {"error": f"Audio analysis failed: {e}"}

    # MFCC
    if mfcc_npy and os.path.exists(mfcc_npy):
        try:
            mfcc = np.load(mfcc_npy)
            summary.update(summarize_vector(mfcc.mean(axis=1), name="mfcc_mean"))
            summary.update(summarize_vector(mfcc.std(axis=1), name="mfcc_std"))
        except Exception as e:
            summary["mfcc_error"] = str(e)

    # wav2vec2
    if wav2vec_npy and os.path.exists(wav2vec_npy):
        try:
            w2v = np.load(wav2vec_npy)
            summary.update(summarize_vector(w2v.mean(axis=0), name="wav2vec2_mean"))
            summary.update(summarize_vector(w2v.std(axis=0), name="wav2vec2_std"))
        except Exception as e:
            summary["wav2vec2_error"] = str(e)

    # prosody
    if prosody_npy and os.path.exists(prosody_npy):
        try:
            pros = np.load(prosody_npy, allow_pickle=True)
            pros_summary = {}
            if isinstance(pros, dict):
                for k, v in pros.items():
                    pros_summary[k] = summarize_vector(v, name=k)
            elif pros.dtype == object and isinstance(pros.item(), dict):
                pros = pros.item()
                for k, v in pros.items():
                    pros_summary[k] = summarize_vector(v, name=k)
            else:
                pros_summary["prosody_array"] = summarize_vector(pros, name="prosody_array")
            summary["prosody_summary"] = pros_summary
        except Exception as e:
            summary["prosody_error"] = str(e)

    return summary


def build_json_for_llm(audio_path, mfcc_npy=None, wav2vec_npy=None, prosody_npy=None):
    summary = build_feature_summary(audio_path, mfcc_npy, wav2vec_npy, prosody_npy)
    return json.dumps(summary, indent=2)

# ======================================================
# PROMPTS (shortened for speed)
# ======================================================
prompt_audio_only = ("""
You are a speech-language pathology analysis model designed for automatic stuttering event detection.

Use ONLY the raw audio clip to detect
stuttering disfluencies (prolongation, block, word repetition, sound repetition).
Provide predictions for all 4 event types - prolongation, block, word repetition, sound repetition and sort them by confidence score.

### STRICT OUTPUT FORMAT
Return ONLY valid JSON in this exact structure (no markdown, commentary, or explanations):

[
  {
    "event_type": "prolongation",
    "start_time": <float or null>,
    "end_time": <float or null>,
    "confidence": <float between 0 and 1>
  },
  {
    "event_type": "block",
    "start_time": <float or null>,
    "end_time": <float or null>,
    "confidence": <float between 0 and 1>
  },
  {
    "event_type": "word repetition",
    "start_time": <float or null>,
    "end_time": <float or null>,
    "confidence": <float between 0 and 1>
  },
  {
    "event_type": "sound repetition",
    "start_time": <float or null>,
    "end_time": <float between 0 and 1>
  }
]

### IMPORTANT:
- Do **not** wrap the JSON in code fences (```json```).
- Do **not** include explanations or commentary.
- Rely solely on the acoustic information from the audio clip.
"""

)

prompt_features_only = ( """
You are a speech-language pathology analysis model designed for automatic stuttering event detection.

Use ONLY the provided acoustic-feature summary (no raw audio) to detect
stuttering disfluencies (prolongation, block, word repetition, sound repetition).
Provide predictions for all 4 event types - prolongation, block, word repetition, sound repetition and sort them by confidence score.

### STRICT OUTPUT FORMAT
Return ONLY valid JSON in this exact structure (no markdown, commentary, or explanations):

[
  {
    "event_type": "prolongation",
    "start_time": <float or null>,
    "end_time": <float or null>,
    "confidence": <float between 0 and 1>
  },
  {
    "event_type": "block",
    "start_time": <float or null>,
    "end_time": <float or null>,
    "confidence": <float between 0 and 1>
  },
  {
    "event_type": "word repetition",
    "start_time": <float or null>,
    "end_time": <float or null>,
    "confidence": <float between 0 and 1>
  },
  {
    "event_type": "sound repetition",
    "start_time": <float or null>,
    "end_time": <float or null>,
    "confidence": <float between 0 and 1>
  }
]

### IMPORTANT:
- Do **not** wrap the JSON in code fences (```json```).
- Do **not** include explanations or commentary.
- Base all decisions only on the numeric and summary data in the feature JSON.
"""

)

#prompt_full = (
#    "You are a speech pathology model. Use both audio and acoustic features to detect "
#    "stuttering disfluencies. Return ONLY JSON output."
#)

PROMPTS = {"audio": prompt_audio_only, "features": prompt_features_only}

# ======================================================
# FILE SELECTION
# ======================================================
random.seed(SEED)
all_wavs = glob(os.path.join(DATA_DIR, "**", "*.wav"), recursive=True)
wav_files = random.sample(all_wavs, min(MAX_SAMPLES, len(all_wavs)))
print(f"Selected {len(wav_files)} audio clips.")

# ======================================================
# UPLOAD AUDIO ONCE
# ======================================================
print("Uploading audio files once (this may take ~1–2 min)...")
uploaded_uris = {}
for w in tqdm(wav_files, desc="Uploading"):
    try:
        up = client.files.upload(file=w)
        uploaded_uris[os.path.basename(w)] = up.uri
    except Exception as e:
        print(f"[Upload error] {w}: {e}")
print(f"Uploaded {len(uploaded_uris)} files successfully.")

# ======================================================
# GEMINI CALL FUNCTION
# ======================================================
def query_gemini(wav_path, mode):
    wav_file = os.path.basename(wav_path)
    base = os.path.splitext(wav_file)[0]
    mfcc_npy = os.path.join(FEATURE_DIR, f"{base}_mfcc.npy")
    w2v_npy = os.path.join(FEATURE_DIR, f"{base}_wav2vec.npy")
    pros_npy = os.path.join(FEATURE_DIR, f"{base}_prosody.npy")
    feature_json = build_json_for_llm(wav_path, mfcc_npy, w2v_npy, pros_npy)

    parts = [{"text": PROMPTS[mode]}]
    if mode in ["audio", "full"]:
        if wav_file not in uploaded_uris:
            raise RuntimeError(f"{wav_file} not uploaded.")
        parts.append({"file_data": {"file_uri": uploaded_uris[wav_file]}})
    if mode in ["features", "full"]:
        parts.append({"text": feature_json})

    # Retry logic for robustness
    for attempt in range(3):
        try:
            resp = client.models.generate_content(model=MODEL, contents=[{"role": "user", "parts": parts}])
            return wav_file, resp.text
        except Exception as e:
            print(f"[Retry {attempt+1}/3] {wav_file}: {e}")
            time.sleep(2)
    return wav_file, ""

# ======================================================
# PARALLEL EXECUTION
# ======================================================
def run_mode(mode):
    print(f"\n=== Running mode: {mode.upper()} ===")
    output_path = os.path.join(OUTPUT_DIR, f"gemini_{mode}_only.csv")
    results = []

    with ThreadPoolExecutor(max_workers=NUM_THREADS) as executor:
        futures = {executor.submit(query_gemini, w, mode): w for w in wav_files}
        for fut in tqdm(as_completed(futures), total=len(futures), desc=f"{mode.upper()}"):
            wav_file, raw_text = fut.result()
            # clean and parse
            json_text = re.sub(r"```[a-zA-Z]*", "", raw_text).replace("```", "").strip()
            split_index = json_text.find("[")
            json_text = json_text[split_index:] if split_index != -1 else "[]"
            try:
                parsed = json.loads(json_text)
                if isinstance(parsed, dict):
                    parsed = [parsed]
            except Exception:
                parsed = []
            for entry in parsed:
                results.append({
                    "file": wav_file,
                    "mode": mode,
                    "event_type": entry.get("event_type"),
                    "start_time": entry.get("start_time"),
                    "end_time": entry.get("end_time"),
                    "confidence": entry.get("confidence")
                })

    pd.DataFrame(results).to_csv(output_path, index=False)
    print(f"Saved {len(results)} predictions → {output_path}")


# ======================================================
# RUN ALL MODES
# ======================================================
for mode in ["audio", "features"]:
    run_mode(mode)

print("\n All modes complete. Results saved in:", OUTPUT_DIR)
````

````python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.metrics import (
    classification_report,
    confusion_matrix,
    ConfusionMatrixDisplay,
    f1_score,
)

# -------------------------------------------------
# CONFIGURATION
# -------------------------------------------------
BASE_DIR = r"C:\Tampu\stuttering"

# absolute paths to all three result files
PRED_FILES = {
    "audio_only": os.path.join(BASE_DIR, "results_fast", "gemini_audio_only.csv"),
    "features_only": os.path.join(BASE_DIR, "results_fast", "gemini_features_only.csv"),
    "full": os.path.join(BASE_DIR, "gemini_prediction_prompt_1.csv"),  # or your full file name
}

ANN_FILE = os.path.join(BASE_DIR, "SEP-28k_joined_labels_episodes.csv")

SAVE_DIR = os.path.join(BASE_DIR, "eval_results_try_2")
os.makedirs(SAVE_DIR, exist_ok=True)


stutter_cols = ["Prolongation", "Block", "SoundRep", "WordRep"]

# -------------------------------------------------
# LOAD ANNOTATIONS
# -------------------------------------------------
#anns = pd.read_csv(ANN_FILE)
#for c in stutter_cols:
#    anns[c] = (anns[c] >= 1).astype(int)

#anns["clip_key"] = (
#    anns["Show"].astype(str) + "_" +
#    anns["EpId"].astype(str) + "_" +
#    anns["ClipId"].astype(str)
#)



# -------------------------------------------------
# LOAD ANNOTATIONS WITH STRICT AGREEMENT
# -------------------------------------------------
anns = pd.read_csv(ANN_FILE)
stutter_cols = ["Prolongation", "Block", "SoundRep", "WordRep"]

anns = pd.read_csv(ANN_FILE)

# keep original 0–3 counts
for c in stutter_cols:
    anns[f"{c}_count"] = anns[c]          # raw 0–3 from CSV
    anns[c] = (anns[f"{c}_count"] >= 3).astype(int)  # strict 3/3 agreement → 1



anns["clip_key"] = (
    anns["Show"].astype(str) + "_" +
    anns["EpId"].astype(str) + "_" +
    anns["ClipId"].astype(str)
)


# -------------------------------------------------
# EVALUATION FUNCTION
# -------------------------------------------------
def evaluate_predictions(pred_csv, tag):
    print(f"\n=== Evaluating {tag} ===")

    preds = pd.read_csv(pred_csv)
    preds["event_type"] = preds["event_type"].str.lower().str.strip()

    # --- collapse duplicates: max confidence per file/event
    preds = (
        preds.groupby(["file", "event_type"], as_index=False)
        .confidence.max()
    )
    preds["clip_key"] = preds["file"].str.replace(".wav", "", regex=False)

    # --- pivot to wide
    pivot = (
        preds.pivot(index="clip_key", columns="event_type", values="confidence")
        .fillna(0.0)
        .reset_index()
    )
    rename_map = {
        "prolongation": "Prolongation",
        "block": "Block",
        "sound repetition": "SoundRep",
        "word repetition": "WordRep",
    }
    pivot.rename(columns=rename_map, inplace=True)

    merged = anns.merge(pivot, on="clip_key", how="inner", suffixes=("_human", "_model"))
    print(f"Merged {len(merged)} clips for evaluation")

    metrics_all = []
    confusion_all = {}

    for col in stutter_cols:
        y_true = merged[f"{col}_human"].astype(int)
        y_conf = merged[f"{col}_model"].astype(float)
        print(f"\n===== Sanity check for {col} =====")
        print("Total clips in anns:", len(anns))
        print("Total positives in anns (>=3):", (anns[col] >= 3).sum())
        
        y_true = merged[f"{col}_human"].astype(int)
        print("Total clips after merge:", len(merged))
        print("Total positives in merged y_true:", y_true.sum())

        # --- threshold sweep for best F1
        thresholds = np.linspace(0.05, 0.95, 19)
        f1s = [f1_score(y_true, (y_conf >= t).astype(int)) for t in thresholds]
        t_best = thresholds[np.argmax(f1s)]

        print(f"\n--- {col} ---")
        print(f"Best threshold: {t_best:.2f}, F1={max(f1s):.3f}")

        # --- metrics @0.5 and @best
        for label, t in [("0.5", 0.5), ("best", t_best)]:
            y_pred = (y_conf >= t).astype(int)
            report = classification_report(y_true, y_pred, output_dict=True, zero_division=0)
            df_report = pd.DataFrame(report).T
            df_report.to_csv(os.path.join(SAVE_DIR, f"class_report_{col}_{tag}_{label}.csv"))

            print(f"\nClassification report ({label} threshold):")
            print(pd.DataFrame(report).T)

            # --- confusion matrix
            # --- confusion matrix
            cm = confusion_matrix(y_true, y_pred)
            
            # extract counts
            tn, fp, fn, tp = cm.ravel()
            
            print(f"\nConfusion Matrix ({col}, threshold={label}):")
            print(cm)
            
            print(f"Counts for {col} @ threshold {label}:")
            print(f"  TP = {tp}")
            print(f"  FP = {fp}")
            print(f"  FN = {fn}")
            print(f"  TN = {tn}\n")
            
            confusion_all[(col, label)] = cm
            
            # plot & save
            disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=["0", "1"])
            disp.plot(cmap="Blues", values_format="d")
            plt.title(f"{col} ({tag}) — threshold={label}")
            plt.tight_layout()
            plt.savefig(
                os.path.join(SAVE_DIR, f"cm_{col}_{tag}_{label}.png"),
                dpi=300, bbox_inches="tight",
            )
            plt.close()

            # --- aggregate metrics
            f1 = report["1"]["f1-score"]
            prec = report["1"]["precision"]
            rec = report["1"]["recall"]
            metrics_all.append({
                "Mode": tag,
                "Event": col,
                "Threshold": label,
                "Best_t": round(t_best, 2),
                "Precision": round(prec, 3),
                "Recall": round(rec, 3),
                "F1": round(f1, 3),
            })

    # --- macro averages
    df_metrics = pd.DataFrame(metrics_all)
    macro = (
        df_metrics.groupby(["Mode", "Threshold"], as_index=False)[["Precision", "Recall", "F1"]]
        .mean()
    )
    df_metrics = pd.concat([df_metrics, macro.assign(Event="Macro Avg")], ignore_index=True)

    out_path = os.path.join(SAVE_DIR, f"metrics_{tag}.csv")
    df_metrics.to_csv(out_path, index=False)
    print(f"\n Saved detailed metrics → {out_path}")
    return df_metrics


# -------------------------------------------------
# RUN FOR ALL 3 MODES
# -------------------------------------------------
all_results = []
for mode, path in PRED_FILES.items():
    df_mode = evaluate_predictions(path, mode)
    all_results.append(df_mode)

# -------------------------------------------------
# COMBINE AND SUMMARY
# -------------------------------------------------
summary = pd.concat(all_results, ignore_index=True)
summary.to_csv(os.path.join(SAVE_DIR, "summary_all_modes.csv"), index=False)
print("\n===== Combined Summary (Macro Averages) =====")
print(summary[summary["Event"] == "Macro Avg"])
print(f" Saved combined summary → {SAVE_DIR}\\summary_all_modes.csv")
````

````python
import pandas as pd
import numpy as np
import os
from sklearn.metrics import confusion_matrix, classification_report, f1_score

# ----------------------------
# Load Ground Truth
# ----------------------------
GT_FILE = r"C:\Users\Nikhi\Downloads\SEP-28k_joined_labels_episodesNEW - SEP-28k_joined_labels_episodesNEW.csv"
gt = pd.read_csv(GT_FILE)

# normalize filename key
gt["file"] = gt["Show"] + "_" + gt["EpId"].astype(str) + "_" + gt["ClipId"].astype(str) + ".wav"
gt = gt[["file", "Prolongation", "Block", "SoundRep", "WordRep"]]

# ----------------------------
# Prediction Files
# ----------------------------
BASE_DIR = r"C:\Tampu\stuttering"
PRED_FILES = {
    "audio_only": os.path.join(BASE_DIR, "results_fast", "gemini_audio_only.csv"),
    "features_only": os.path.join(BASE_DIR, "results_fast", "gemini_features_only.csv"),
    "full": os.path.join(BASE_DIR, "gemini_prediction_prompt_1.csv"),
}

# ----------------------------
# Label Mapping
# ----------------------------
EVENT_MAP = {
    "block": "Block",
    "prolongation": "Prolongation",
    "sound repetition": "SoundRep",
    "word repetition": "WordRep",
}

# ----------------------------
# Helper: Load prediction + keep raw confidence
# ----------------------------
def load_pred_file(pred_path):
    df = pd.read_csv(pred_path)

    # Wide format, keep raw confidence scores
    pred = df.pivot_table(
        index="file",
        columns="event_type",
        values="confidence",
        aggfunc="max"
    ).fillna(0)

    pred = pred.rename(columns=EVENT_MAP)

    # Ensure all 4 columns exist
    for col in ["Prolongation", "Block", "SoundRep", "WordRep"]:
        if col not in pred.columns:
            pred[col] = 0

    pred.reset_index(inplace=True)
    return pred


# ----------------------------
# NEW: Best threshold search (0.00–1.00)
# ----------------------------
def compute_best_threshold(y_true, y_scores):
    thresholds = np.linspace(0, 1, 101)
    best_f1 = -1
    best_t = 0.5

    for t in thresholds:
        y_pred = (y_scores >= t).astype(int)
        f1 = f1_score(y_true, y_pred, zero_division=0)

        if f1 > best_f1:
            best_f1 = f1
            best_t = t

    return best_t, best_f1


# ----------------------------
# Evaluation Function
# ----------------------------
def evaluate_predictions(gt, pred, model_name="model"):
    print("\n" + "=" * 100)
    print(f" RESULTS FOR: {model_name}")
    print("=" * 100)

    merged = pd.merge(gt, pred, on="file", how="inner", suffixes=("_gt", "_score"))

    for label in ["Prolongation", "Block", "SoundRep", "WordRep"]:
        y_true = merged[f"{label}_gt"]
        y_scores = merged[f"{label}_score"]

        # default threshold predictions
        y_pred_05 = (y_scores >= 0.5).astype(int)

        # find best threshold
        best_t, best_f1 = compute_best_threshold(y_true, y_scores)

        print(f"\n--------------------------------------------------")
        print(f"Label: {label}")
        print(f"Best threshold = {best_t:.2f}")
        print(f"Best F1 = {best_f1:.4f}")

        # performance at threshold = 0.5
        print(f"\nF1 at threshold 0.5 = {f1_score(y_true, y_pred_05):.4f}")

        print("\nConfusion Matrix @ 0.5:")
        print(confusion_matrix(y_true, y_pred_05))

        print("\nClassification Report @ 0.5:")
        print(classification_report(y_true, y_pred_05, digits=4))


# ----------------------------
# Run evaluation
# ----------------------------
for name, path in PRED_FILES.items():
    pred = load_pred_file(path)
    evaluate_predictions(gt, pred, name)

````
